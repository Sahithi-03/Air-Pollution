# Advection-Diffusion

Advection diffusion equation has two methods to solve

$$\partial_t c + u \partial_x c = D \partial_{xx} c$$


With the following boundary conditions

\begin{align}
    C(x,0) &= \delta(x) \\
    C(\pm L, t) &= 0 \\
\end{align}


For the advection diffusion problem we will have $\mathrm{Pe} \gg 1$

$\mathrm{Pe} = \frac{uL}{\mu} = \frac{advection}{diffucsion}$ is the peclet number and when $\mathrm{Pe} \gg 1$

$u$ is flow velocity

$L$ is length scale

$D$ is diffusivity
